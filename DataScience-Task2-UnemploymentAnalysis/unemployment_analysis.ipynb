```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display all columns
pd.set_option('display.max_columns', None)

# Set visualization style
sns.set_style("whitegrid")

print("Libraries imported successfully!")
```

```python
# Load unemployment datasets
df1 = pd.read_csv("Unemployment in India.csv")
df2 = pd.read_csv("Unemployment_Rate_upto_11_2020.csv")

print("Datasets loaded successfully!")
```

```python

# Inspect Dataset 1
print("========== DATASET 1 ==========")

print("Shape:", df1.shape)

print("\nColumn Names:")
print(df1.columns.tolist())

print("\nFirst 5 Rows:")
display(df1.head())

print("\nMissing Values:")
print(df1.isnull().sum())


# Inspect Dataset 2
print("\n========== DATASET 2 ==========")

print("Shape:", df2.shape)

print("\nColumn Names:")
print(df2.columns.tolist())

print("\nFirst 5 Rows:")
display(df2.head())

print("\nMissing Values:")
print(df2.isnull().sum())

```

```python
# Check date columns in both datasets

print("Dataset 1 date information:")
print(df1[' Date'].head())

print("\nDataset 2 date information:")
print(df2[' Date'].head())
```

```python

# Remove unwanted spaces from column names

df1.columns = df1.columns.str.strip()
df2.columns = df2.columns.str.strip()

# Convert Date columns into datetime format

df1['Date'] = pd.to_datetime(df1['Date'], dayfirst=True, errors='coerce')
df2['Date'] = pd.to_datetime(df2['Date'], dayfirst=True, errors='coerce')

# Display updated information

print("Dataset 1 cleaned successfully!")
print(df1.dtypes)

print("\nDataset 2 cleaned successfully!")
print(df2.dtypes)
```

```python

# Check missing values before cleaning

print("Dataset 1 Missing Values:")
print(df1.isnull().sum())

print("\nDataset 2 Missing Values:")
print(df2.isnull().sum())
```

```python

# Check completely empty rows

print("Dataset 1 completely empty rows:",
      df1.isnull().all(axis=1).sum())

print("Dataset 2 completely empty rows:",
      df2.isnull().all(axis=1).sum())

print("\nDataset 1 Shape:", df1.shape)
print("Dataset 2 Shape:", df2.shape)
```

```python

# Remove completely empty rows from Dataset 1

df1 = df1.dropna(how='all')

# Reset index
df1 = df1.reset_index(drop=True)

# Verify cleaning
print("Dataset 1 cleaned successfully!")

print("\nNew Dataset Shape:", df1.shape)

print("\nMissing Values After Cleaning:")
print(df1.isnull().sum())

print("\nFirst 5 Rows:")
display(df1.head())
```

```python

# Check duplicate records

print("Dataset 1 Duplicate Rows:", df1.duplicated().sum())
print("Dataset 2 Duplicate Rows:", df2.duplicated().sum())
```

```python

# Statistical Analysis of Unemployment Rate

print("UNEMPLOYMENT RATE ANALYSIS")

print("\nAverage Unemployment Rate:")
print(df1['Estimated Unemployment Rate (%)'].mean())

print("\nMaximum Unemployment Rate:")
print(df1['Estimated Unemployment Rate (%)'].max())

print("\nMinimum Unemployment Rate:")
print(df1['Estimated Unemployment Rate (%)'].min())

print("\nOverall Statistical Summary:")
display(df1['Estimated Unemployment Rate (%)'].describe())
```

```python

# Calculate average unemployment rate by region

state_unemployment = df1.groupby('Region')[
    'Estimated Unemployment Rate (%)'
].mean().sort_values(ascending=False)

print("Average Unemployment Rate by State:")
display(state_unemployment)
```

```python

plt.figure(figsize=(12, 7))

state_unemployment.plot(kind='bar')

plt.title('Average Unemployment Rate by State in India')
plt.xlabel('States')
plt.ylabel('Average Unemployment Rate (%)')
plt.xticks(rotation=90)
plt.tight_layout()

plt.show()
```

```python

# Monthly Unemployment Trend Analysis

monthly_unemployment = df1.groupby('Date')[
    'Estimated Unemployment Rate (%)'
].mean().sort_index()

print("Monthly Unemployment Rates:")
display(monthly_unemployment)

# Plot the trend
plt.figure(figsize=(12, 6))

plt.plot(
    monthly_unemployment.index,
    monthly_unemployment.values,
    marker='o',
    linewidth=2
)

plt.title('Monthly Unemployment Rate in India')
plt.xlabel('Date')
plt.ylabel('Unemployment Rate (%)')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()

plt.show()
```

```python

# Employment Analysis

average_employment = df1.groupby('Region')[
    'Estimated Employed'
].mean().sort_values(ascending=False)

print("Average Employment by State:")
display(average_employment)

# Visualize employment
plt.figure(figsize=(12, 6))

average_employment.plot(kind='bar')

plt.title('Average Estimated Employment by State')
plt.xlabel('State')
plt.ylabel('Estimated Employed')
plt.xticks(rotation=90)
plt.tight_layout()

plt.show()
```

```python

# COVID-19 Impact Analysis

# Separate pre-COVID and COVID periods
df1['Period'] = df1['Date'].apply(
    lambda date: 'Pre-COVID' if date < pd.Timestamp('2020-03-01')
    else 'COVID Period'
)

# Calculate average unemployment rate
covid_analysis = df1.groupby('Period')[
    'Estimated Unemployment Rate (%)'
].mean()

print("COVID-19 Impact on Unemployment:")
display(covid_analysis)

# Visualize comparison
plt.figure(figsize=(8, 5))

covid_analysis.plot(
    kind='bar',
    title='Unemployment Rate: Pre-COVID vs COVID Period'
)

plt.xlabel('Period')
plt.ylabel('Average Unemployment Rate (%)')
plt.xticks(rotation=0)
plt.tight_layout()

plt.show()
```

```python

# Distribution of Unemployment Rate

plt.figure(figsize=(9, 5))

sns.histplot(
    df1['Estimated Unemployment Rate (%)'],
    bins=20,
    kde=True
)

plt.title('Distribution of Unemployment Rates in India')
plt.xlabel('Unemployment Rate (%)')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()
```

```python

# Correlation Heatmap

numeric_data = df1.select_dtypes(include=np.number)

plt.figure(figsize=(9, 6))

sns.heatmap(
    numeric_data.corr(),
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap of Unemployment Data')
plt.tight_layout()

plt.show()
```

```python

# State-wise unemployment trends

selected_states = ['Karnataka', 'Maharashtra', 'Tamil Nadu']

state_data = df1[df1['Region'].isin(selected_states)]

plt.figure(figsize=(12, 6))

sns.lineplot(
    data=state_data,
    x='Date',
    y='Estimated Unemployment Rate (%)',
    hue='Region',
    marker='o'
)

plt.title('Unemployment Trends in Selected Indian States')
plt.xlabel('Date')
plt.ylabel('Unemployment Rate (%)')
plt.xticks(rotation=45)
plt.legend(title='State')
plt.tight_layout()

plt.show()
```

```python

# Top 10 states with highest average unemployment

top_10 = state_unemployment.head(10)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=top_10.values,
    y=top_10.index
)

plt.title('Top 10 States with Highest Average Unemployment')
plt.xlabel('Average Unemployment Rate (%)')
plt.ylabel('State')

plt.tight_layout()
plt.show()
```

```python

# Final Summary

print("========== PROJECT SUMMARY ==========")

print("Total Records:", len(df1))

print("Number of States:", df1['Region'].nunique())

print(
    "Average Unemployment Rate:",
    round(df1['Estimated Unemployment Rate (%)'].mean(), 2), "%"
)

print(
    "Highest Unemployment State:",
    state_unemployment.idxmax()
)

print(
    "Lowest Unemployment State:",
    state_unemployment.idxmin()
)

print("\nCOVID-19 Comparison:")
display(covid_analysis)
```
